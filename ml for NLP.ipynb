```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

path = r"/content/E-commerce.xlsx"

df = pd.read_excel(path)

print(df.head())

x = df["Comment"]
y = df["Tag"]

x = x.fillna("")
y = y.fillna("")

x = x.astype(str)

pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        tokenizer=lambda text: text.split(),
        token_pattern=None
    )),

    ("model", LogisticRegression(max_iter=1000))
])

kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

y_pred = cross_val_predict(
    pipeline,
    x,
    y,
    cv=kfold
)

print("Accuracy:", accuracy_score(y, y_pred))

print("Precision:",
      precision_score(y, y_pred, average='weighted'))

print("Recall:",
      recall_score(y, y_pred, average='weighted'))

print("F1 Score:",
      f1_score(y, y_pred, average='weighted'))

print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
```

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)

path = r"/content/E-commerce.xlsx"

df = pd.read_excel(path)

x = df["Comment"].fillna("").astype(str)
y = df["Tag"].fillna("").astype(str)

pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        tokenizer=lambda text: text.split(),
        token_pattern=None
    )),

    ("model", LogisticRegression(
        max_iter=1000
    ))
])

kfold = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

y_pred = cross_val_predict(
    pipeline,
    x,
    y,
    cv=kfold
)

print("Accuracy:", accuracy_score(y, y_pred))

print("Precision:",
      precision_score(y, y_pred, average="weighted"))

print("Recall:",
      recall_score(y, y_pred, average="weighted"))

print("F1 Score:",
      f1_score(y, y_pred, average="weighted"))

print("\nClassification Report:\n")
print(classification_report(y, y_pred))

cm = confusion_matrix(y, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Load dataset
df = pd.read_excel("E-commerce.xlsx")

# Input and Output
X = df["Comment"].astype(str)
y = df["Tag"].fillna("").astype(str) # Added .fillna("").astype(str) to handle NaNs

# Convert text to vectors
tfidf = TfidfVectorizer(max_features=5000)

X_tfidf = tfidf.fit_transform(X)

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42
)

# Model
model = LogisticRegression()

# Train
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Load dataset
df = pd.read_excel("E-commerce.xlsx")

# Input and Output
X = df["Comment"].astype(str)
y = df["Tag"].fillna("").astype(str)  # Handle NaN values in the target variable

# TF-IDF
tfidf = TfidfVectorizer(max_features=5000)

X_tfidf = tfidf.fit_transform(X)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42
)

# SVC Model
svc_model = SVC(kernel='linear')

# Train
svc_model.fit(X_train, y_train)

# Prediction
y_pred = svc_model.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Load dataset
df = pd.read_excel("E-commerce.xlsx")

# Input and Output
X = df["Comment"].astype(str)
y = df["Tag"].fillna("").astype(str) # Handle NaN values in the target variable

# TF-IDF Vectorization
tfidf = TfidfVectorizer(max_features=5000)

X_tfidf = tfidf.fit_transform(X)

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42
)

# Random Forest Model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train
rf_model.fit(X_train, y_train)

# Prediction
y_pred = rf_model.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
